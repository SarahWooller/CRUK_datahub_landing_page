```python
import sys
from pathlib import Path
import markdown
from weasyprint import HTML, CSS
```

```python
# Minimal CSS to format the output cleanly
DEFAULT_CSS = """
@page {
    size: A4;
    margin: 20mm;
}
body {
    font-family: -apple-system, BlinkMacSystemFont, "Segoe UI", Roboto, sans-serif;
    line-height: 1.6;
    color: #333;
}
h1, h2, h3 {
    color: #111;
}
code {
    background-color: #f4f4f4;
    padding: 2px 4px;
    border-radius: 4px;
    font-family: monospace;
}
pre {
    background-color: #f4f4f4;
    padding: 10px;
    border-radius: 4px;
    overflow-x: auto;
}
blockquote {
    border-left: 4px solid #ccc;
    margin: 0;
    padding-left: 16px;
    color: #666;
}
table {
    border-collapse: collapse;
    width: 100%;
}
th, td {
    border: 1px solid #ddd;
    padding: 8px;
    text-align: left;
}
"""

def md_to_pdf(input_md_path: str, output_pdf_path: str = None) -> None:
    input_path = Path(input_md_path)
    
    if not input_path.exists():
        raise FileNotFoundError(f"Input file not found: {input_md_path}")
        
    if output_pdf_path is None:
        output_pdf_path = input_path.with_suffix(".pdf")
    else:
        output_pdf_path = Path(output_pdf_path)

    # Read Markdown content
    with open(input_path, "r", encoding="utf-8") as f:
        md_content = f.read()

    # Convert Markdown to HTML (enabling tables and code highlighting extensions)
    html_content = markdown.markdown(
        md_content,
        extensions=["tables", "fenced_code", "codehilite", "toc"]
    )

    # Wrap in standard HTML structure
    full_html = f"""
    <!DOCTYPE html>
    <html>
    <head>
        <meta charset="utf-8">
    </head>
    <body>
        {html_content}
    </body>
    </html>
    """

    # Render HTML to PDF
    HTML(string=full_html).write_pdf(
        target=output_pdf_path,
        stylesheets=[CSS(string=DEFAULT_CSS)]
    )
    
    print(f"Successfully converted '{input_path}' to '{output_pdf_path}'")
```

```python
md_to_pdf("")
```
